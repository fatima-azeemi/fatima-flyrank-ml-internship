## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

# Simple threshold rules (like "Flag all pages older than 180 days" or "Flag any page with less than 1,000 words") fail because SEO data is non-linear and context-dependent. A short 300-word FAQ page might perform exceptionally well, while a massive 4,000-word blog post might decay rapidly due to outdated content structure. Machine Learning models find complex multi-dimensional interactions across dozens of metrics simultaneously and adapt dynamically when search landscapes shift, whereas fixed rules break quickly.
```
